````python
employees = [
    {"name": "John", "salary": 50000},
    {"name": "David", "salary": 70000},
    {"name": "Alice", "salary": 80000}
]

employee_names = []
def get_employee_names(employees):
  for employee in employees:
    employee_names.append(employee.get('name'))
  return(employee_names)

print(get_employee_names(employees))
````

````python
#Exercise 1
employees = [
    {"name": "John", "salary": 50000},
    {"name": "David", "salary": 70000},
    {"name": "Alice", "salary": 80000}
]

high_salary_employees = []

def highSalaryEmployees(employees):
  for employee in employees:
    if employee.get("salary") >= 70000:
      high_salary_employees.append(employee.get('name'))
  return high_salary_employees

print(highSalaryEmployees(employees))
````

````python
#Exercise 2 Count employees missing salary.
employees = [
    {"name": "John"},
    {"name": "David", "salary": 70000},
    {"name": "Alice"},
    {"name": "Bob", "salary": 80000}
]

count = 0

def missing_salary_count(employee):
  return employee.get("salary") is None


for employee in employees:
  if missing_salary_count(employee):
    count += 1
print(f"Missing Salary Employees: {count}")
````

````python
#Senior Engineer Challenge
valid_records = 0
invalid_records = 0
missing_salary = []

with open('employees.csv') as file:
  for line in file:
    parts = line.strip().split(',')
    if not parts[1]:
        invalid_records += 1
        missing_salary.append(parts[0])
    else:
        valid_records += 1
result = {
    "valid_records": valid_records,
    "invalid_records": invalid_records,
    "missing_salary": missing_salary
}

print(result)
````


````python
#Catch Specific Errors
##ValueError

try:
  number = int('abc')
  print(number)
except ValueError:
  print('Invalid Number')
````

````python
#Data Engineering Challenge
salaries = [
    "50000",
    "abc",
    "70000"
]

result = []

for salary in salaries:
  try:
    result.append(int(salary))
  except ValueError:
    print('invalid Value')
print(result)
````

````python
#Senior Engineer Challenge

employees = [
    {"name": "John", "salary": "50000"},
    {"name": "David", "salary": "abc"},
    {"name": "Alice", "salary": "80000"}
]

valid_records = 0
invalid_records = 0
invalid_names = []

for employee in employees:
  try:
    int(employee.get('salary'))
    valid_records += 1

  except ValueError:
      invalid_records += 1
      invalid_names.append(employee['name'])

result = {
    "valid_records": valid_records,
    "invalid_records": invalid_records,
    "invalid_names": invalid_names
}
print(result)
````

#Day 18 – List Comprehensions 🔥

````python
#Traditional Loop
numbers = [1,2,3,4]

result = []
for number in numbers:
  result.append(number * number)

print(result)
````

````python
#[expression for item in iterable]
squre = [number*number for number in numbers]

print(squre)
````

````python
salaries = [50000, 70000, 90000]
result = []

for salary in salaries:
  if salary >= 70000:
    result.append(salary)
print(result)
````

````python
result = [salary for salary in salaries if salary >= 70000]
print(result)
````

````python
#Pattern 1 – Extract Data
employees = [
    {"name": "John"},
    {"name": "David"},
    {"name": "Alice"}
]

names = [employee.get('name') for employee in employees]
print(names)
````

````python
#Pattern 2 – Filter Data
employees = [
    {"name": "John", "salary": 50000},
    {"name": "David", "salary": 80000},
    {"name": "Alice", "salary": 90000}
]

high_salary = [employee['name'] for employee in employees if employee.get('salary') >= 70000]
print(high_salary)
````

#Dictionary Comprehension

Structure:
```
{
    key: value
    for item in iterable
}
```



````python
names = ["John", "David", "Alice"]

result = {name: len(name) for name in names}
print(result)
````

````python
#Exercise 1 – High Salary Names
employees = [
    {"name": "John", "salary": 50000},
    {"name": "David", "salary": 80000},
    {"name": "Alice", "salary": 90000},
    {"name": "Bob", "salary": 40000}
]

result = [employee.get('name') for employee in employees if employee.get('salary') >= 70000]
print(result)
````

````python
#Exercise 3 – Extract All Skills
employees = [
    {"name": "John", "skills": ["Python", "SQL"]},
    {"name": "Alice", "skills": ["Excel", "Power BI"]}
]

skills_list = [skill for employee in employees for skill in employee["skills"]]
print(skills_list)
````
