#Day 23 – Encapsulation (Protecting Object Data) 🔥

What is Encapsulation?

Definition:

    Encapsulation means keeping data and the methods that work on that data together inside a class, while controlling how the data is accessed or modified.

Simple meaning:

    Don't allow anyone to change your object's data in any way they want.

````python
#Without Encapsulation
class Employee:

    def __init__(self, name, salary):
        self.name = name
        self.salary = salary

employee = Employee("John", 50000) #Now anyone can do:

employee.salary = -10000 #❌ Invalid salary. Nothing prevents it.

print(employee.salary)
````

````python
#With Encapsulation

#Setter Method
class Employee:
  def __init__(self, name, salary):
    self.name = name
    self.salary = salary

  def set_salary(self, salary):
    if salary >= 0:
      self.salary = salary

employee = Employee('Nirmal', 5000)
employee.set_salary(55000)

print(employee.salary)
````

````python
#Senior Engineer Challenge
class Employee:
  def __init__(self, name, salary, department):
    self.__name = name
    self.__salary = salary
    self.__department = department

  def get_name(self):
    return self.__name

  def get_salary(self):
    return self.__salary

  def get_department(self):
    return self.__department

  def set_salary(self, salary):
    if salary >= 0:
      self.__salary = salary

  def set_department(self, department):
    if department is not None:
      self.__department = department

  def has_salary(self):
    return self.__salary is not None

  def has_department(self):
    return self.__department is not None

  def is_valid(self):
    return self.has_salary() and self.has_department()

employee = Employee(
    "John",
    50000,
    "IT"
)

print(employee.is_valid())
employee.set_salary(-500)
print(employee.get_salary())
````

#Day 24 – Polymorphism & Method Overriding 🔥

**What is Polymorphism?**

**Definition:**

  **Polymorphism means:**

    One method name
    Different behavior

The same method can behave differently depending on the object.

````python
#Without Polymorphism
class Employee:

    def work(self):
        print("Working")
````

````python
#With Polymorphism
class Employee:

    def work(self):
        print("Working")

class Manager(Employee):

    def work(self):
        print("Managing Team")

class Developer(Employee):

    def work(self):
        print("Writing Code")
````

````python
employees = [
    Employee(),
    Manager(),
    Developer()
]

for employee in employees:
    employee.work()
````

Same method.

Different outputs.

This is Polymorphism.

Method Overriding

    When a child class creates a method with the same name as the parent:

````python
#Parent Class
class Employee:
  def work(self):
    print("Working")
#Child Class
class Manager(Employee): #The child replaces (overrides) the parent's version.
  def work(self):
      print("Managing Team")

employees = [Employee(), Manager()]

for employee in employees:
  employee.work()
````

````python
class Employee:
  def work(self):
    print("Employee Working")

class Manager(Employee):
  def work(self):
    super().work()
    print("Managing Team")

employees = [Employee(), Manager()]

for employee in employees:
  employee.work()
````

````python
#Method Overriding
#Base class
class Animal:
  def speak(self):
    pass

#Derived Class
class Dog(Animal):
  def speak(self):
    return "WooW!"

class Cat(Animal):
  def speak(self):
    return "Meow!"

class Cow(Animal):
  def speak(self):
    return "Moo!"

def animal_sound(animal):
  print(animal.speak())

dog = Dog()
cat = Cat()
cow = Cow()

animal_sound(dog)
animal_sound(cat)
animal_sound(cow)
````

#Day 25 – Abstraction (The 4th Pillar of OOP) 🔥

What is Abstraction?

Definition:

Abstraction means:

    Hide implementation details.
    Show only what is necessary.

Simple meaning:

    User knows WHAT to do,
    not HOW it is done.

image.png

````python
#Exercise 1
from abc import ABC, abstractmethod

class Animal(ABC):

    @abstractmethod
    def sound(self):
        pass

class Dog(Animal):

    def sound(self):
        return "Bark"

dog = Dog()

print(dog.sound())
````

````python
#Exercise 2
from abc import ABC, abstractmethod

class Animal(ABC):

    @abstractmethod
    def sound(self):
        pass

animal = Animal()

'''Answer: Animal is an abstract class.

It contains an abstract method sound().

Python does not allow creating an object of an abstract class until all abstract methods are implemented.

Therefore:

animal = Animal()

raises TypeError.'''
````

````python
#Senior Engineer Challenge
from abc import ABC, abstractmethod

class Validator(ABC):
  @abstractmethod
  def validate(self, employee):
    pass

class SalaryValidator(Validator):
  def validate(self, employee):
    return employee.get("salary") is not None

class DepartmentValidator(Validator):
  def validate(self, employee):
    return employee.get("department") is not None

class ValidationEngine:
  def run(self, employee, validators):
    passed = 0
    failed = 0
    for validator in validators:
      if validator.validate(employee):
        passed += 1
      else:
        failed += 1

    return {
            "passed": passed,
            "failed": failed
        }


employee1 = {
    "name": "John",
    "salary": 50000,
    "department": "IT"
}

validators = [SalaryValidator(), DepartmentValidator()]

engine = ValidationEngine()
result1 = engine.run(employee1, validators)

print(result1)


employee2 = {
    "name": "David",
    "salary": None,
    "department": None
}

result2 = engine.run(employee2, validators)
print(result2)
````

````python
class Employee:
  def __init__(self, name, salary, department, joining_date):
    self.name = name
    self.salary = salary
    self.department = department
    self.joining_date = joining_date

from datetime import datetime

class JoiningDateValidator:
  def validate(self, employee):
    joining_date = datetime.strptime(employee.joining_date, '%Y-%m-%d')
    today = datetime.today().replace(hour=0, minute=0, second=0)
    return joining_date <= today

class SalaryValidator:
  def validate(self, employee):
    return employee.salary is not None and employee.salary > 0

class DepartmentValidator:
    def validate(self, employee):
        # For demo: assume always valid
        return employee.department is not None

class ValidationEngine:
  def __init__(self):
    self.validators = [SalaryValidator(), DepartmentValidator(), JoiningDateValidator()]
  def run(self, employee):
    passed = 0
    failed = 0
    for validator in self.validators:
      if validator.validate(employee):
        passed += 1
      else:
        failed += 1
    return {
            "passed": passed,
            "failed": failed}


# Create Employee
employee = Employee(
    name="John",
    salary=50000,
    department="PS",
    joining_date="2024-05-20"
)

engine = ValidationEngine()
result = engine.run(employee)

print(result)
````

Methods in re Module
To find a pattern we use different set of re character sets that allows to search for a match in a string.

1. re.match(): searches only in the beginning of the first line of the string and returns matched objects if found, else returns None.
2. re.search: Returns a match object if there is one anywhere in the string, including multiline strings.
3. re.findall: Returns a list containing all matches
4. re.split: Takes a string, splits it at the match points, returns a list
5. re.sub: Replaces one or many matches within a string

````python
import pandas as pd

class EmployeeCleaner:
    def __init__(self, df):
        self.df = df

    def fill_missing_salary(self):
        # Fill missing salary with average salary
        avg_salary = self.df["salary"].mean()
        self.df["salary"] = self.df["salary"].fillna(avg_salary)
        return avg_salary

    def high_salary(self):
        # Employees with salary greater than 50000
        return self.df[self.df["salary"] > 50000]

    def it_department(self):
        # Employees in IT department
        return self.df[self.df["department"] == "IT"]

    def employee_count(self):
        # Total number of employees
        return len(self.df)


# Input
employees = {
    "name": ["John", "David", "Alice", "Bob"],
    "salary": [50000, None, 90000, 40000],
    "department": ["IT", "HR", "Finance", "IT"]
}

df = pd.DataFrame(employees)

# Create object
emp = EmployeeCleaner(df)

# Output
print("Average salary:", emp.fill_missing_salary())
print("\nHigh salary employees:")
print(emp.high_salary())

print("\nIT employees:")
print(emp.it_department())

print("\nEmployee count:", emp.employee_count())
````

````python
import pandas as pd

class EmployeeCleaner:
    def __init__(self, df):
        self.df = df

    def department_average(self):
        return self.df.groupby('department')['salary'].mean()

    def department_max(self):
        return self.df.groupby('department')['salary'].max()

    def department_employee_count(self):
        return self.df.groupby('department')['name'].count()

    def department_summary(self):
      return self.df.groupby("department")["salary"].agg(
        ["count", "mean", "max", "min"])


# Input
employees = {
    "name": ["John", "David", "Alice", "Bob", "Sam", "Kevin"],
    "salary": [50000, 70000, 90000, 40000, 60000, 80000],
    "department": ["IT", "HR", "Finance", "IT", "HR", "Finance"]
}

df = pd.DataFrame(employees)

emp = EmployeeCleaner(df)

print(emp.department_average())
print(emp.department_employee_count())
print(emp.department_max())
print(emp.department_summary())
````
