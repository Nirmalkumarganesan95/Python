```python
def annual_salary_calculation(employee_name, monthly_salary):
  annual_salary = monthly_salary * 12
  return annual_salary

employee_name = input("Enter Name: ")
monthly_salary = int(input("Enter Salary"))

annual_salary_calculation(employee_name, monthly_salary)
```

```python
def area_calculations(length, width):
  area = length * width
  return area

length = int(input("Enter length"))
width = int(input("Enter width"))

area_calculations(length, width)
```

```python
def gross_salary_calculation(employee_name, monthly_salary, bonus):
  gross_salary = (monthly_salary * 12) + bonus
  print(f"Employee: {employee_name}")
  print(f"Annual Salary: {gross_salary}")

employee_name = input("Enter Name: ")
monthly_salary = int(input("Enter Salary: "))
bonus = int(input("Enter bonus: "))

gross_salary_calculation(employee_name, monthly_salary, bonus)
```
