```python
#Print number of salaries
#Sort salaries
#Print smallest salary
#Print largest salary

salaries = [70000, 50000, 90000, 60000]


print(len(salaries))

salaries.sort()
print(salaries)

smallest_salary = salaries[0]
print(f"smallest salary is: {smallest_salary}")

largest_salary = salaries[-1]
print(f"largest salary is: {largest_salary}")
```
